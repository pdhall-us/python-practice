```python
f=open(r"C:\Users\Prateek\Desktop\file.txt", "w") # Creating file object using open function: Syntax: open(path, mode, encoding)
data="This is the new file created in python."
f.write(data) # writing data into file using write method
f.close() # closing the file
```

```python
# Reading the file
f=open(r"C:\Users\Prateek\Desktop\file.txt","r")
data=f.read()
print(data)
f.close()
```

```python
# Appending data into existing file.
f=open(r"C:\Users\Prateek\Desktop\file.txt","a")
data="\nThis is line added through append mode."
f.write(data)
f.close()
```

```python
f=open(r"C:\Users\Prateek\Desktop\file.txt","r")
data=f.readlines()
print(data)
f.close()
```
