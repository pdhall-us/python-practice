#### Program for 2 highest values in dictionary

```python
#### Expected Output:
#### B -> 8000
#### D -> 7000
```

```python
# Simple with 2-loops

sales = {
    'A': 5000,
    'B': 8000,
    'C': 3000,
    'D': 9000
}
first=0
second=0
first_name=""
second_name=""
for i in sales:
    if sales[i]>first:
        first=sales[i]
        first_name=i
for i in sales:
    if sales[i]>second and i!=first_name:
        second=sales[i]
        second_name=i
        
print(f"{first_name} -> {first}")
print(f"{second_name} -> {second}")
```

```python
# With single loop

sales = {
    'A': 5000,
    'B': 8000,
    'C': 3000,
    'D': 9000
}

first=0
second=0
first_name=""
second_name=""

for i in sales:
    if sales[i]>first:
        second=first
        second_name=first_name

        first=sales[i]
        first_name=i

    elif sales[i]>second:
        second=sales[i]
        second_name=i
```
