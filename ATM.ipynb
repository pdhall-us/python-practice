```python

def bal():
    global bank_bal
    print(f"Your current balance: {bank_bal}")

def deposit():
    global bank_bal
    d=int(input("Enter Deposit Amount: "))
    bank_bal=bank_bal+d
    print(f"Amount Deposited: {d}")
    bal()

def withdraw():
    global bank_bal
    w=int(input("Enter Withdrawal Amount: "))
    if bank_bal<w:
        print(f"Bank bal: {bank_bal} not enough for withdrawal of {w}")
    else:
        bank_bal=bank_bal-w
        print(f"Amount Withdrawn: {w}")
        bal()

bank_bal=0
print(f"Your current Bank balance is {bank_bal}, please deposit!!!")
deposit()

while True:
    print(f"\nEnter your choice: Your account has {bank_bal}.\na. Deposit\nb. Withdrwal\nc. Current balance\nd. Exit\n")
    choice=input().lower()

    if choice=='a':
        deposit()
    elif choice=='b':
        withdraw()
    elif choice=='c':
        bal()
    elif choice=='d':
        break
    else:
        print("\nEnter a valid choice!!")
```
