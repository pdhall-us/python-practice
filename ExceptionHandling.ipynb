## Exception Handling

- Exception handling is a technique for managing errors gracefully instead of allowing application to crash unexpectedly.

#### 4-types of blocks in exception handling

| Blocks | Description |
| --- | --- |
| try | This block contains that part of code where error can occur. |
| exception | Code inside this block executes when any error occurs in try block. |
| else | This block get executed if there is no error in try block |
| finally | This block will get executed irrespective of that if any error occurs or not. |

```python
# Suppose if there is simple addition program and if you pass string to it, this will return ValueError.

num1=int(input("Num1: "))
num2=int(input("Num2: "))
num3=num1+num2

print(f"Sum: {num3}")
```

```python
# We can resolve it using exception handling and guide the user why the error occured and what you should do.

try:
    num1=int(input("Num1: "))
    num2=int(input("Num2: "))
except:
    print("Please enter integer value")
else:
    num3=num1+num2
    print(f"Sum: {num3}")
```

```python
# what if we have to handle 2 errors, value error and zero division error.
# If there is more than one type of errors, you can manage it as follow.

try:
    num1=int(input("Num1: "))
    num2=int(input("Num2: "))
    num3=num1/num2
except ValueError: # Here we are handling value error, if input is not integer.
    print("Please enter integer value")
except ZeroDivisionError: # Here we are handling zero division error, if denominator is zero.
    print("Please don't use 0 as denominator")
else:
    print(f"Quotient: {num3}")
finally: # This block executes irrespective if error occurs or not
    print("Thank you for visiting!!!")
```

#### Finally limitation
- Only limitation is that, when we use recursion then it will still get executed and stored in cache and displayed in continuous manner at last.

```python
def add():
    try:
        num1=int(input("Enter number: "))
        num2=int(input("Enter number: "))
    except:
        print("Please enter only integer value")
        add()
    else:
        num3=num1+num2
        print("Add: ",num3)
    finally:
        print("Thank you")
add()
```

- As we can see above the clear example of the limitation during recursion.
